```python
pro_matches_summary_df = pd.read_parquet(f"{silver_ds_dir}/pro_matches_summary.parquet")
#print(pro_matches_summary_df.head())
#pro_matches_summary_df.columns

patches_stats = (
    pro_matches_summary_df
    .groupby('patch')
    .agg(
        matches=('gameid', 'nunique'),
        players=('playername', 'nunique'),
        champions=('champion', 'nunique'),
        avg_gamelength=('gamelength', 'mean'),
        avg_player_kills=('kills', 'mean'),
        avg_player_deaths=('deaths', 'mean'),
        avg_player_assists=('assists', 'mean')
    )
    .assign(
        match_type='professional'
    )
    .reset_index()
)
#print(patches_stats)

matches_per_patch = pro_matches_summary_df.groupby('patch')['gameid'].nunique()

pick_per_patch = pro_matches_summary_df.groupby(['patch', 'champion'])['gameid'].nunique()
#print(picked_per_patch)

pick_rate = (
    pick_per_patch
    .div(matches_per_patch, level='patch')
    .rename('pick_rate')
    .reset_index()
) 
#print(pick_rate)

# champions_stats = (
#     pick_rate
#     .merge(ban_rate_df, on=['patch', 'champion'])
# )
# #print(champions_stats.loc[champions_stats['champion'] == 'Jinx'])

top_pick_per_patch = (
    pick_rate
    .loc[pick_rate.groupby('patch')['pick_rate'].idxmax()]
    .rename(columns={"champion": "top_pick_champion", "pick_rate": "top_pick_rate"})
    .reset_index(drop=True)
)

top_ban_per_patch = (
    ban_rate_df
    .loc[ban_rate_df.groupby('patch')['ban_rate'].idxmax()]
    .rename(columns={"champion": "top_ban_champion", "ban_rate": "top_ban_rate"})
    .reset_index(drop=True)
)
# print(top_pick_per_patch)
# print(top_ban_per_patch)

top_champions_stats = (
    top_pick_per_patch
    .merge(top_ban_per_patch, on='patch')
)


pro_patches_stats = (
    patches_stats
    .merge(top_champions_stats, on='patch')
)
print(pro_patches_stats.head())

# print('Min patch in 1: ', patches_stats['patch'].min(), 'Max patch in 1: ', patches_stats['patch'].max())
# print('Min patch in 2: ', patches_stats2['patch'].min(), 'Max patch in 2: ', patches_stats2['patch'].max())
# print('Min patch in banlist: ', ban_rate_df['patch'].min(), 'Max patch in banlist: ', ban_rate_df['patch'].max())
# print('Min patch in picklist: ', pick_rate['patch'].min(), 'Max patch in picklist: ', pick_rate['patch'].max())
# print('Min patch in top champions: ', top_champions_stats['patch'].min(), 'Max patch in top champions: ', top_champions_stats['patch'].max())
```
